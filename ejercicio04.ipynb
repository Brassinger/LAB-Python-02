### a. Palabras de más de 4 letras

```python
palabras = ["sol", "perro", "casa", "universidad"]
resultado = list(filter(lambda palabra: len(palabra) > 4, palabras))
print(resultado)
```

### b. Conservar elementos nulos (None)

```python
datos = [5, None, "hola", None, 8]
resultado = list(filter(lambda x: x is None, datos))
print(resultado)
```

### c. Palabras que empiezan por vocal

```python
palabras = ["avion", "casa", "Elefante", "oso", "mesa"]
resultado = list(filter(lambda palabra: palabra[0].lower() in "aeiou", palabras))
print(resultado)
```

### d. Palíndromos

```python
palabras = ["oso", "casa", "reconocer", "radar", "python"]
resultado = list(filter(lambda palabra: palabra.lower() == palabra[::-1].lower(), palabras))
print(resultado)
```

### e. Números que terminan en 5

```python
numeros = [15, 22, 35, 40, 55, 91]
resultado = list(filter(lambda n: abs(n) % 10 == 5, numeros))
print(resultado)
```

### f. Productos de precio mayor a 100

```python
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]
resultado = list(filter(lambda producto: producto["precio"] > 100, productos))
print(resultado)
```
