### a. Filtrar pares y elevarlos al cuadrado

```python
numeros = [5, 2, 8, 3, 4, 7]
resultado = sorted([n ** 2 for n in numeros if n % 2 == 0])
print(resultado)
```

### b. Cubos mayores que 100 y convertirlos a cadenas (interpretación del texto cortado)

```python
numeros = [2, 3, 4, 5, 6]
cubos = [n ** 3 for n in numeros]
resultado = [str(n) for n in sorted(cubos) if n > 100]
print(resultado)
```

### c. Filtrar >10, elevar al cuadrado y ordenar de mayor a menor (interpretación del texto cortado)

```python
numeros = [5, 12, 3, 15, 20, 7]
resultado = sorted([n ** 2 for n in numeros if n > 10], reverse=True)
print(resultado)
```
