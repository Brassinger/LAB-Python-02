# Ejercicio 3 – Mapa (map)

Laboratorio 02. Cada apartado se ejecuta en una celda independiente de Google Colab.

### a. Celsius a Fahrenheit

```python
temperaturas = [0, 20, 25, 100]
resultado = list(map(lambda c: c * 9 / 5 + 32, temperaturas))
print(resultado)
```

### b. Sumar dos listas por posiciones

```python
lista_a = [1, 2, 3]
lista_b = [10, 20, 30]
resultado = list(map(lambda a, b: a + b, lista_a, lista_b))
print(resultado)
```

### c. Áreas de círculos

```python
from math import pi
radios = [1, 2, 3, 4]
resultado = list(map(lambda r: pi * r ** 2, radios))
print(resultado)
```

### d. Aumentar precios 10 %

```python
precios = {"pan": 1000, "leche": 2500, "café": 5000}
resultado = list(map(lambda par: (par[0], par[1] * 1.10), precios.items()))
print(resultado)
```

### e. Multiplicar una matriz por 10

```python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
resultado = list(map(lambda fila: list(map(lambda x: x * 10, fila)), matriz))
print(resultado)
```
