### a. Tuplas (n, n²)

```python
numeros = [1, 2, 3, 4, 5]
resultado = [(n, n ** 2) for n in numeros]
print(resultado)
```

### b. Cubos de números mayores que 10

```python
numeros = [2, 5, 11, 12, 15]
resultado = [n ** 3 for n in numeros if n > 10]
print(resultado)
```

### c. Filtrar números pares

```python
numeros = [1, 2, 3, 4, 5, 6]
resultado = [n for n in numeros if n % 2 == 0]
print(resultado)
```

### d. Convertir números en textos valor: ...

```python
numeros = [10, 15, 20]
resultado = ["valor: " + str(n) for n in numeros]
print(resultado)
```

### e. Convertir Celsius a Fahrenheit

```python
temperaturas = [0, 20, 25, 100]
resultado = [(c * 9 / 5) + 32 for c in temperaturas]
print(resultado)
```

### f. Palabras de más de 5 letras

```python
palabras = ["gato", "universidad", "casa", "python", "sol"]
resultado = [palabra for palabra in palabras if len(palabra) > 5]
print(resultado)
```

### g. Iniciales de las palabras

```python
palabras = ["Ana", "Juan", "Maria", "Pedro"]
resultado = [palabra[0] for palabra in palabras]
print(resultado)
```

### h. Indicar par o impar

```python
numeros = [1, 2, 3, 4, 5]
resultado = ["par" if n % 2 == 0 else "impar" for n in numeros]
print(resultado)
```

### i. Aplanar una matriz

```python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
resultado = [n for fila in matriz for n in fila]
print(resultado)
```
