### Juego completo. La persona empieza. Puede retirar 1, 2 o 3 piedras; pierde quien no puede mover. MAX es la computadora, MIN la persona.

```python
def minimax(piedras, turno_cpu):
    """Evaluar el juego: +1 gana CPU, -1 gana humano."""
    if piedras == 0:
        if turno_cpu:
            return -1
        return 1

    valores = []
    for quitar in [1, 2, 3]:
        if quitar <= piedras:
            valor = minimax(piedras - quitar, not turno_cpu)
            valores.append(valor)

    if turno_cpu:
        return max(valores)
    return min(valores)


def turno_computadora(piedras):
    """Elegir la jugada con el mejor resultado para MAX."""
    mejor_movimiento = 1
    mejor_valor = -2

    for quitar in [1, 2, 3]:
        if quitar <= piedras:
            valor = minimax(piedras - quitar, False)
            if valor > mejor_valor:
                mejor_valor = valor
                mejor_movimiento = quitar

    return mejor_movimiento


def jugar():
    """Simular NIM con humano primero y CPU después."""
    piedras = 10
    print("Comienza el juego con", piedras, "piedras")

    while piedras > 0:
        print("Piedras disponibles:", piedras)
        try:
            quitar = int(input("¿Cuántas piedras quitas (1, 2 o 3)? "))
        except ValueError:
            print("Ingresa un número válido.")
            continue

        if quitar < 1 or quitar > 3 or quitar > piedras:
            print("Movimiento no permitido.")
            continue

        piedras -= quitar
        if piedras == 0:
            print("¡Ganaste!")
            break

        quitar_cpu = turno_computadora(piedras)
        piedras -= quitar_cpu
        print("La computadora quitó", quitar_cpu, "piedras")

        if piedras == 0:
            print("Ganó la computadora")


jugar()
```
